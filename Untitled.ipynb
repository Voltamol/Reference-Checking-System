```python
import math

def burgerPurchase(numBurgers):
    patties_per_pack = 10
    buns_per_pack = 8

    # Calculate the number of packages needed
    patties_packs = math.ceil(numBurgers / patties_per_pack)
    buns_packs = math.ceil(numBurgers / buns_per_pack)

    return (patties_packs, buns_packs)
```

```python
print(burgerPurchase(50))  # Output: (5, 7)
print(burgerPurchase(10))  # Output: (1, 2)
print(burgerPurchase(5))   # Output: (1, 1)
print(burgerPurchase(0))   # Output: (0, 0)
```

```python
def oldest(d1, m1, y1, d2, m2, y2):
    if y1 < y2:
        return 1
    elif y2 < y1:
        return 2
    elif m1 < m2:
        return 1
    elif m2 < m1:
        return 2
    elif d1 < d2:
        return 1
    elif d2 < d1:
        return 2
    else:
        return 0
```

```python
print(oldest(10, 5, 1990, 15, 5, 1990))  # Output: 1
print(oldest(20, 3, 1985, 15, 6, 1985))  # Output: 2
print(oldest(5, 12, 2000, 5, 12, 2000))  # Output: 0
```
