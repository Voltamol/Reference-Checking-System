```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
#model = tf.keras.models.load_model('NiceAdmin/flask/models/comment_classifier_model.h5')
model = tf.keras.models.load_model('comment_classifier_model.h5')
```

```python
def preprocess_input(user_input):
    #-----------------------------------------------------------
    max_words = 10000
    max_len = 100
    comments=user_input
    #-----------------------------------------------------------
    tokenizer = Tokenizer(num_words=max_words)
    tokenizer.fit_on_texts(comments)
    sequences = tokenizer.texts_to_sequences(comments)
    padded_sequences = pad_sequences(sequences, maxlen=max_len)
    #-----------------------------------------------------------
    return padded_sequences
```

```python
def make_predictions(response):
    data=preprocess_input(response)
    sentiment = model.predict(data)
    return sentiment
```

```python
def classify_sentiment(user_input):
    predictions=make_predictions(user_input)
    index = np.argmax(predictions, axis=1)
    # Map predicted classes to sentiment labels
    sentiment_labels = ['positive', 'negative', 'neutral']
    predicted_sentiments = [sentiment_labels[i] for i in index]
    #most_common_sentiment = mode(predicted_sentiments)
    #return most_common_sentiment
    total_labels=len(predicted_sentiments)
    classes={label:0 for label in sentiment_labels}
    for prediction in predicted_sentiments:
        classes[prediction]+=1
    for label in classes:
        classes[label]=100*classes[label]/total_labels
    return classes
```

```python
comment:str="""
I have worked with the young man for quite some time now under my company, he is very talented when it comes to programming and data analytics.
he has the ability to learn quickly and hist problem solving skills are on point. 
he loves taking on challenges and usually comes up with creative solutions to problems.
However he also has a few weaknesses particularly on showing up on time 
and does not regard the rule of formal dressing which is mandatory during weekdays
there is much room for improvement though and a greater possibility of digital transformation in your company
"""
```

```python
classify_sentiment(comment)
```
